```python
# ============================================================
# Independent Closed-loop vs Open-loop Comparison Experiment (includes all necessary class definitions)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter, defaultdict
import os
import re
import random
import time
import glob
import warnings
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from datetime import datetime, timedelta
from typing import List, Dict, Tuple, Any, Set
import langdetect
from langdetect import detect, DetectorFactory
import json
import networkx as nx

warnings.filterwarnings('ignore')
DetectorFactory.seed = 0

# Check TextBlob
try:
    from textblob import TextBlob
    TEXTBLOB_AVAILABLE = True
except ImportError:
    TEXTBLOB_AVAILABLE = False
    print("TextBlob not installed, using dictionary-based sentiment analysis")


# ============================================================
# 1. Smart Large Model Wrapper (copied from your original code)
# ============================================================

class SmartModelWrapper:
    """Smart large model wrapper - learns user style, no template generation"""
    
    def __init__(self, model_path="C:/Users/fs/Desktop/lwz/models/MiniCPM-4.0-8B"):
        self.model_path = model_path
        self.tokenizer = None
        self.model = None
        self.device = None
        self._load_model()

    def _load_model(self):
        try:
            print(f"Loading model: {self.model_path}")
            self.tokenizer = AutoTokenizer.from_pretrained(self.model_path, trust_remote_code=True)
    
            if torch.cuda.is_available():
                self.device = torch.device("cuda:0")
                print(f"Using GPU: {torch.cuda.get_device_name(0)}")
                self.model = AutoModelForCausalLM.from_pretrained(
                    self.model_path,
                    torch_dtype=torch.float16,
                    device_map={"": 0},
                    trust_remote_code=True,
                    low_cpu_mem_usage=True
                )
            else:
                self.device = torch.device("cpu")
                print("Using CPU")
                self.model = AutoModelForCausalLM.from_pretrained(
                    self.model_path,
                    torch_dtype=torch.float32,
                    device_map=None,
                    trust_remote_code=True
                ).to(self.device)
            
            print("Model loaded successfully")
        except Exception as e:
            print(f"Model loading failed: {e}")
            raise
    
    def analyze_user_style(self, user_posts: List[str]) -> Dict:
        """Deeply analyze user style"""
        if not user_posts:
            return {
                'language': 'en',
                'style': 'neutral',
                'common_words': [],
                'sentence_patterns': [],
                'emotional_tone': 'neutral',
                'sample_phrases': [],
                'post_samples': []
            }
        
        language = self._detect_language(user_posts)
        common_words = self._extract_common_words(user_posts, language)
        sentence_patterns = self._analyze_sentence_patterns(user_posts)
        emotional_tone = self._analyze_emotional_tone(user_posts)
        sample_phrases = self._extract_sample_phrases(user_posts)
        style_desc = self._build_style_description(language, sentence_patterns, emotional_tone)
        
        return {
            'language': language,
            'style': style_desc,
            'common_words': common_words[:10],
            'sentence_patterns': sentence_patterns,
            'emotional_tone': emotional_tone,
            'sample_phrases': sample_phrases[:5],
            'post_samples': user_posts[:3]
        }
    
    def _detect_language(self, posts: List[str]) -> str:
        if not posts:
            return 'en'
        lang_counts = {}
        for post in posts[:10]:
            if isinstance(post, str) and len(post.strip()) > 10:
                try:
                    lang = detect(post)
                    lang_counts[lang] = lang_counts.get(lang, 0) + 1
                except:
                    continue
        if lang_counts:
            return max(lang_counts.items(), key=lambda x: x[1])[0]
        return 'en'
    
    def _extract_common_words(self, posts: List[str], language: str) -> List[str]:
        if not posts:
            return []
        all_text = ' '.join(posts).lower()
        if language in ['ja', 'zh-cn', 'zh-tw', 'ko']:
            words = re.findall(r'[\u4e00-\u9fff\u3040-\u309f\u30a0-\u30ff\uac00-\ud7af]{2,}', all_text)
        else:
            words = re.findall(r'\b[a-zA-Z]{3,}\b', all_text)
        word_freq = Counter(words)
        return [word for word, _ in word_freq.most_common(20)]
    
    def _analyze_sentence_patterns(self, posts: List[str]) -> List[str]:
        patterns = []
        for post in posts[:10]:
            if isinstance(post, str):
                if '?' in post:
                    patterns.append('questioning')
                if '!' in post:
                    patterns.append('exclamatory')
                if '...' in post or '…' in post:
                    patterns.append('pausing')
                if len(post.split()) < 10:
                    patterns.append('concise')
                elif len(post.split()) > 25:
                    patterns.append('elaborate')
        pattern_counts = Counter(patterns)
        return [pattern for pattern, _ in pattern_counts.most_common(3)]
    
    def _analyze_emotional_tone(self, posts: List[str]) -> str:
        if not posts:
            return 'neutral'
        positive_words = ['good', 'great', 'excellent', 'positive', 'optimistic', 'happy', 'love', 'like']
        negative_words = ['bad', 'poor', 'negative', 'problem', 'issue', 'hate', 'dislike']
        pos_count, neg_count = 0, 0
        for post in posts[:10]:
            if isinstance(post, str):
                text_lower = post.lower()
                pos_count += sum(1 for word in positive_words if word in text_lower)
                neg_count += sum(1 for word in negative_words if word in text_lower)
        if pos_count > neg_count * 2:
            return 'positive'
        elif neg_count > pos_count * 2:
            return 'negative'
        else:
            return 'neutral'
    
    def _extract_sample_phrases(self, posts: List[str]) -> List[str]:
        phrases = []
        for post in posts[:10]:
            if isinstance(post, str):
                sentences = re.split(r'[.!?]+', post)
                for sentence in sentences:
                    words = sentence.strip().split()
                    if 5 <= len(words) <= 15:
                        phrases.append(sentence.strip())
        return phrases[:10]
    
    def _build_style_description(self, language: str, patterns: List[str], tone: str) -> str:
        style_parts = []
        lang_map = {'en': 'English', 'ja': 'Japanese', 'zh-cn': 'Chinese', 'ko': 'Korean'}
        lang_desc = lang_map.get(language, language)
        style_parts.append(f"writes in {lang_desc}")
        if patterns:
            pattern_desc = ', '.join(patterns[:2])
            style_parts.append(f"uses {pattern_desc} sentences")
        if tone == 'positive':
            style_parts.append("generally positive tone")
        elif tone == 'negative':
            style_parts.append("often critical tone")
        return ', '.join(style_parts)
    
    def generate_context_aware_post(self, user_style: Dict, network_context: Dict, 
                                   positivity_boost: float = 0.5) -> str:
        """Generate environment-aware post"""
        if self.model is None or self.tokenizer is None:
            raise ValueError("Large model not loaded, cannot generate content")
        
        prompt = self._build_intelligent_prompt(user_style, network_context, positivity_boost)
        
        try:
            inputs = self.tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512)
            model_device = self.device
            if hasattr(self.model, 'device'):
                model_device = self.model.device
            inputs = {k: v.to(model_device) for k, v in inputs.items()}
            
            generation_config = {
                'max_new_tokens': 128,
                'temperature': 0.8,
                'do_sample': True,
                'top_p': 0.9,
                'repetition_penalty': 1.1,
                'pad_token_id': self.tokenizer.eos_token_id,
                'num_return_sequences': 1
            }
            
            with torch.no_grad():
                outputs = self.model.generate(**inputs, **generation_config)
            
            generated_text = self.tokenizer.decode(outputs[0], skip_special_tokens=True)
            result = self._extract_response(generated_text, prompt)
            return result
            
        except Exception as e:
            print(f"Large model generation failed: {e}")
            language = user_style.get('language', 'en')
            if language == 'zh-cn':
                return "This discussion on this topic is quite interesting."
            elif language == 'ja':
                return "この話題についての議論は興味深いです。"
            else:
                return "Interesting discussion on this topic."
    
    def _build_intelligent_prompt(self, user_style: Dict, network_context: Dict, 
                                 positivity_boost: float) -> str:
        user_desc = f"""User Writing Style Analysis:
- Primary language: {user_style.get('language', 'en')}
- Style characteristics: {user_style.get('style', 'neutral')}
- Common words/phrases: {', '.join(user_style.get('common_words', [])[:5])}
- Sentence patterns: {', '.join(user_style.get('sentence_patterns', []))}
- Emotional tone: {user_style.get('emotional_tone', 'neutral')}

Recent user posts:
{chr(10).join(['- ' + post for post in user_style.get('post_samples', [])[:2]])}"""
        
        env_desc = "Current Network Environment:\n"
        if network_context.get('sentiment'):
            env_desc += f"- Overall sentiment: {network_context['sentiment']}\n"
        if network_context.get('trending_words'):
            trending = network_context['trending_words'][:5]
            env_desc += f"- Trending words: {', '.join(trending)}\n"
        if network_context.get('recent_topics'):
            topics = network_context['recent_topics'][:3]
            env_desc += f"- Recent topics: {', '.join(topics)}\n"
        if network_context.get('consensus_analysis'):
            env_desc += f"- Network consensus: {network_context['consensus_analysis']}\n"
        
        strategy = "Generation Strategy:\n"
        if positivity_boost > 0.7:
            strategy += "- Use strong positive language, show enthusiasm\n"
        elif positivity_boost > 0.4:
            strategy += "- Use moderately positive language\n"
        else:
            strategy += "- Keep tone balanced, slightly positive\n"
        
        prompt = f"""You are an AI assistant simulating social media posts. Generate a post matching the user's style while responding to the network environment.

{user_desc}

{env_desc}

{strategy}

CRITICAL REQUIREMENTS:
1. Write in the EXACT same language as the user's recent posts
2. Match the user's writing style, vocabulary, and sentence patterns
3. Reference the network environment appropriately
4. Length: 1-2 sentences

Generate the post exactly as the user would write it:"""
        return prompt
    
    def _extract_response(self, generated_text: str, prompt: str) -> str:
        if generated_text.startswith(prompt):
            response = generated_text[len(prompt):].strip()
        else:
            lines = generated_text.split('\n')
            response = lines[-1].strip() if lines else generated_text
        response = re.sub(r'\s+', ' ', response).strip()
        response = re.sub(r'^"(.*)"$', r'\1', response)
        response = re.sub(r'^(As|So|Well|Actually|Basically),?\s*', '', response, flags=re.IGNORECASE)
        return response


# ============================================================
# 2. Time-Aware Particle Class
# ============================================================

class TimeAwareParticle:
    """Time-aware random walk particle"""
    
    def __init__(self, particle_id: int, start_node: str, network: nx.Graph,
                 user_data: Dict[str, List[Tuple[str, datetime]]],
                 max_steps: int = 8, time_decay_factor: float = 0.1):
        self.particle_id = particle_id
        self.current_node = start_node
        self.network = network
        self.user_data = user_data
        self.max_steps = max_steps
        self.time_decay_factor = time_decay_factor
        self.visited_nodes = [start_node]
        self.collected_data = []
        self.collected_posts = []
        self.step_count = 0
        self.current_time = self._get_current_time()
        self._collect_node_data(start_node)
    
    def _get_current_time(self) -> datetime:
        all_times = []
        for posts_with_time in self.user_data.values():
            for _, post_time in posts_with_time:
                if post_time:
                    all_times.append(post_time)
        if all_times:
            return max(all_times)
        else:
            return datetime.now()
    
    def _get_time_weight(self, post_time: datetime) -> float:
        if not post_time:
            return 0.1
        time_diff_hours = (self.current_time - post_time).total_seconds() / 3600
        weight = np.exp(-self.time_decay_factor * time_diff_hours)
        return max(0.1, min(1.0, weight))
    
    def _collect_node_data(self, node: str):
        if node not in self.user_data:
            return
        posts_with_time = self.user_data[node]
        weighted_posts = []
        for post_content, post_time in posts_with_time[:5]:
            weight = self._get_time_weight(post_time)
            weighted_posts.append((post_content, post_time, weight))
        weighted_posts.sort(key=lambda x: x[2], reverse=True)
        for post_content, post_time, _ in weighted_posts[:3]:
            self.collected_data.append((post_content, post_time))
            self.collected_posts.append(post_content)
    
    def step(self) -> bool:
        if self.step_count >= self.max_steps:
            return False
        neighbors = list(self.network.neighbors(self.current_node))
        if not neighbors:
            return False
        next_node = self._time_aware_selection(neighbors)
        self.current_node = next_node
        self.visited_nodes.append(next_node)
        self._collect_node_data(next_node)
        self.step_count += 1
        return self.step_count < self.max_steps
    
    def _time_aware_selection(self, neighbors: List[str]) -> str:
        selection_scores = []
        for neighbor in neighbors:
            base_score = 0.1
            if self.network.has_edge(self.current_node, neighbor):
                base_score = self.network[self.current_node][neighbor].get('weight', 0.1)
            time_score = 0.0
            if neighbor in self.user_data and self.user_data[neighbor]:
                latest_time = None
                for _, post_time in self.user_data[neighbor]:
                    if post_time:
                        if latest_time is None or post_time > latest_time:
                            latest_time = post_time
                if latest_time:
                    time_score = self._get_time_weight(latest_time) * 0.3
            penalty = -0.2 if neighbor in self.visited_nodes else 0.0
            total_score = base_score + time_score + random.uniform(0, 0.05) + penalty
            selection_scores.append(max(0.01, total_score))
        total_score = sum(selection_scores)
        if total_score > 0:
            probabilities = [score / total_score for score in selection_scores]
            return random.choices(neighbors, weights=probabilities, k=1)[0]
        else:
            return random.choice(neighbors)
    
    def get_collected_posts(self) -> List[str]:
        return self.collected_posts


# ============================================================
# 3. Intelligent SCL Method (Closed-loop)
# ============================================================

class IntelligentSCLMethod:
    """Intelligent SCL method - deep user style learning and environment integration"""
    
    def __init__(self, train_df, model_wrapper, positivity_boost: float = 0.5,
                 n_particles: int = 5, max_steps_per_particle: int = 6,
                 recent_hours: int = 48, time_decay_factor: float = 0.1):
        self.train_df = train_df
        self.model = model_wrapper
        self.positivity_boost = positivity_boost
        self.n_particles = n_particles
        self.max_steps_per_particle = max_steps_per_particle
        self.recent_hours = recent_hours
        self.time_decay_factor = time_decay_factor
        
        self.train_df = self._parse_timestamps(train_df)
        self.recent_data = self._get_recent_data(recent_hours)
        self.network = self._build_recent_social_network()
        self.user_data_with_time = self._prepare_user_data_with_time()
        self.network_context = self._explore_recent_network_environment()
        self.user_profiles = self._extract_user_profiles()
    
    def _parse_timestamps(self, df):
        df_copy = df.copy()
        if 'timeStamp' in df_copy.columns:
            try:
                df_copy['parsed_time'] = pd.to_datetime(df_copy['timeStamp'], format='%Y-%m-%d %H:%M:%S')
            except:
                df_copy['parsed_time'] = pd.date_range(end=pd.Timestamp.now(), periods=len(df_copy), freq='h')
        else:
            df_copy['parsed_time'] = pd.date_range(end=pd.Timestamp.now(), periods=len(df_copy), freq='h')
        return df_copy.sort_values('parsed_time', ascending=True).reset_index(drop=True)
    
    def _get_recent_data(self, recent_hours: int):
        if 'parsed_time' not in self.train_df.columns:
            return self.train_df
        latest_time = self.train_df['parsed_time'].max()
        time_threshold = latest_time - pd.Timedelta(hours=recent_hours)
        return self.train_df[self.train_df['parsed_time'] >= time_threshold].copy()
    
    def _build_recent_social_network(self) -> nx.Graph:
        G = nx.Graph()
        if len(self.recent_data) > 0:
            user_counts = self.recent_data['userName'].value_counts()
            recent_users = user_counts.index[:min(20, len(user_counts))]
        else:
            user_counts = self.train_df['userName'].value_counts()
            recent_users = user_counts.index[:min(20, len(user_counts))]
        for user in recent_users:
            G.add_node(user)
        for i, user1 in enumerate(recent_users):
            user1_posts = self.recent_data[self.recent_data['userName'] == user1]['text'].dropna().tolist()[:5]
            for user2 in recent_users[i+1:]:
                user2_posts = self.recent_data[self.recent_data['userName'] == user2]['text'].dropna().tolist()[:5]
                if user1_posts and user2_posts:
                    similarity = self._calculate_content_similarity(user1_posts, user2_posts)
                    if similarity > 0.2:
                        G.add_edge(user1, user2, weight=similarity)
        return G
    
    def _calculate_content_similarity(self, posts1: List[str], posts2: List[str]) -> float:
        if not posts1 or not posts2:
            return 0.0
        def extract_keywords(posts):
            all_text = ' '.join([str(p) for p in posts])
            words = re.findall(r'\b[a-zA-Z]{3,}\b', all_text.lower())
            return set(words)
        keywords1, keywords2 = extract_keywords(posts1), extract_keywords(posts2)
        if not keywords1 or not keywords2:
            return 0.0
        return len(keywords1 & keywords2) / len(keywords1 | keywords2) if (keywords1 | keywords2) else 0.0
    
    def _prepare_user_data_with_time(self) -> Dict[str, List[Tuple[str, datetime]]]:
        user_data = {}
        if len(self.recent_data) > 0:
            recent_users = self.recent_data['userName'].value_counts().index[:min(15, len(self.recent_data))]
        else:
            recent_users = self.train_df['userName'].value_counts().index[:min(15, len(self.train_df))]
        for user in recent_users:
            user_df = self.train_df[self.train_df['userName'] == user]
            user_df_sorted = user_df.sort_values('parsed_time', ascending=False)
            posts_with_time = []
            for _, row in user_df_sorted.iterrows():
                if pd.notna(row['text']):
                    posts_with_time.append((str(row['text']), row['parsed_time']))
            if posts_with_time:
                user_data[user] = posts_with_time[:8]
        return user_data
    
    def _explore_recent_network_environment(self) -> Dict:
        active_users = list(self.user_data_with_time.keys())
        if not active_users:
            return self._get_default_network_context()
        all_collected_posts = []
        for i in range(self.n_particles):
            start_user = random.choice(active_users)
            particle = TimeAwareParticle(
                particle_id=i,
                start_node=start_user,
                network=self.network,
                user_data=self.user_data_with_time,
                max_steps=self.max_steps_per_particle,
                time_decay_factor=self.time_decay_factor
            )
            for _ in range(self.max_steps_per_particle):
                if not particle.step():
                    break
            all_collected_posts.extend(particle.get_collected_posts())
        if len(all_collected_posts) < 10:
            recent_posts = self.recent_data['text'].dropna().tolist()[:20]
            all_collected_posts.extend(recent_posts)
        return self._analyze_collected_data(all_collected_posts)
    
    def _analyze_collected_data(self, collected_posts: List[str]) -> Dict:
        if not collected_posts:
            return self._get_default_network_context()
        return {
            'sentiment': self._analyze_sentiment(collected_posts),
            'trending_words': self._extract_trending_words(collected_posts),
            'recent_topics': self._extract_topics(collected_posts),
            'consensus_analysis': self._analyze_consensus(collected_posts),
            'time_frame': f'last {self.recent_hours} hours'
        }
    
    def _extract_trending_words(self, posts: List[str]) -> List[str]:
        if not posts:
            return []
        all_text = ' '.join([str(p) for p in posts[:30] if isinstance(p, str)])
        words = re.findall(r'\b[a-zA-Z]{4,}\b', all_text.lower())
        stopwords = {'the', 'and', 'you', 'for', 'are', 'this', 'that', 'with', 'have', 'from'}
        filtered_words = [w for w in words if w.lower() not in stopwords]
        word_freq = Counter(filtered_words)
        return [word for word, _ in word_freq.most_common(15) if word_freq[word] >= 2]
    
    def _analyze_sentiment(self, posts: List[str]) -> str:
        if not posts:
            return "neutral"
        positive_count, negative_count = 0, 0
        for post in posts[:20]:
            if isinstance(post, str):
                text_lower = post.lower()
                pos_words = ['good', 'great', 'excellent', 'positive', 'optimistic', 'happy']
                neg_words = ['bad', 'poor', 'negative', 'problem', 'issue', 'worry']
                pos_count = sum(1 for word in pos_words if word in text_lower)
                neg_count = sum(1 for word in neg_words if word in text_lower)
                if pos_count > neg_count:
                    positive_count += 1
                elif neg_count > pos_count:
                    negative_count += 1
        total = positive_count + negative_count
        if total > 0:
            pos_ratio = positive_count / total
            if pos_ratio > 0.6:
                return "generally positive"
            elif pos_ratio > 0.4:
                return "slightly positive"
            elif pos_ratio < 0.2:
                return "generally negative"
            elif pos_ratio < 0.4:
                return "slightly negative"
        return "mixed/neutral"
    
    def _extract_topics(self, posts: List[str]) -> List[str]:
        if not posts:
            return []
        topics = []
        for post in posts[:15]:
            if isinstance(post, str):
                noun_phrases = re.findall(r'\b([A-Z][a-z]+(?:\s+[A-Z][a-z]+)*)\b', post)
                topics.extend([np for np in noun_phrases if len(np.split()) <= 3])
        topic_freq = Counter(topics)
        return [topic for topic, _ in topic_freq.most_common(10) if topic_freq[topic] >= 2]
    
    def _analyze_consensus(self, posts: List[str]) -> str:
        sentiment = self._analyze_sentiment(posts)
        if "positive" in sentiment:
            return "moderate positive consensus"
        elif "negative" in sentiment:
            return "moderate negative consensus"
        else:
            return "mixed opinions"
    
    def _get_default_network_context(self) -> Dict:
        return {
            'sentiment': 'neutral',
            'trending_words': [],
            'recent_topics': [],
            'consensus_analysis': 'mixed opinions',
            'time_frame': 'unknown'
        }
    
    def _extract_user_profiles(self):
        profiles = {}
        user_counts = self.train_df['userName'].value_counts()
        users = user_counts.index[:min(10, len(user_counts))]
        for user in users:
            user_posts = self.train_df[self.train_df['userName'] == user]['text'].dropna().tolist()
            if not user_posts:
                continue
            try:
                user_style = self.model.analyze_user_style(user_posts)
                profiles[user] = user_style
            except:
                profiles[user] = self._simple_user_analysis(user_posts)
        return profiles
    
    def _simple_user_analysis(self, user_posts: List[str]) -> Dict:
        language = 'en'
        try:
            sample_text = ' '.join(user_posts[:3])
            if len(sample_text) > 10:
                language = detect(sample_text)
        except:
            pass
        all_text = ' '.join(user_posts).lower()
        words = re.findall(r'\b[a-zA-Z]{4,}\b', all_text)
        word_freq = Counter(words)
        return {
            'language': language,
            'style': f'user writing in {language}',
            'common_words': [word for word, _ in word_freq.most_common(10)],
            'sentence_patterns': ['declarative'],
            'emotional_tone': 'neutral',
            'sample_phrases': user_posts[:2],
            'post_samples': user_posts[:3]
        }
    
    def generate_for_user(self, user_name: str, num_posts: int = 1) -> List[str]:
        if user_name not in self.user_profiles:
            user_posts = self.train_df[self.train_df['userName'] == user_name]['text'].dropna().tolist()
            if not user_posts:
                raise ValueError(f"User {user_name} has no historical post data")
            user_style = self.model.analyze_user_style(user_posts)
        else:
            user_style = self.user_profiles[user_name]
        
        try:
            post = self.model.generate_context_aware_post(
                user_style, self.network_context, self.positivity_boost
            )
            return [post]
        except Exception as e:
            print(f"Generation failed: {e}")
            return [f"Generated content for {user_name}."]


# ============================================================
# 4. Open-loop Variant
# ============================================================

class IntelligentSCLMethod_Open(IntelligentSCLMethod):
    """Open-loop version: fixed environment, no feedback"""
    
    def __init__(self, train_df, model_wrapper, positivity_boost=0.5,
                 n_particles=5, max_steps_per_particle=6,
                 recent_hours=48, time_decay_factor=0.1):
        super().__init__(train_df, model_wrapper, positivity_boost,
                         n_particles, max_steps_per_particle,
                         recent_hours, time_decay_factor)
        # Save initial environment, do not update afterwards
        self._frozen_context = self.network_context.copy()
    
    def _explore_recent_network_environment(self):
        """Always return the initially detected environment"""
        if hasattr(self, '_frozen_context'):
            return self._frozen_context
        else:
            return super()._explore_recent_network_environment()


# ============================================================
# 5. Evaluation Metrics Class
# ============================================================

class ImprovedEvaluationMetrics:
    """Improved evaluation metrics class"""
    
    def __init__(self, train_df, test_df):
        self.train_df = train_df
        self.test_df = test_df
        self.use_textblob = TEXTBLOB_AVAILABLE
        self.env_features = self._extract_environment_features()
    
    def _extract_environment_features(self):
        recent_posts = self.train_df['text'].dropna().tolist()
        if len(recent_posts) > 100:
            recent_posts = recent_posts[-100:]
        if not recent_posts:
            return {'trending_words': [], 'common_topics': [], 'avg_sentiment': 0.5}
        
        all_text = ' '.join([str(p) for p in recent_posts])
        words = re.findall(r'\b[a-zA-Z]{3,}\b', all_text.lower())
        word_freq = Counter(words)
        stopwords = {'the', 'and', 'you', 'for', 'are', 'this', 'that', 'with', 'have', 'from'}
        trending_words = [word for word, _ in word_freq.most_common(30) 
                         if word not in stopwords and word_freq[word] >= 2]
        
        avg_sentiment = 0.5
        if self.use_textblob:
            try:
                sentiments = []
                for post in recent_posts[:20]:
                    if isinstance(post, str) and len(post) > 10:
                        blob = TextBlob(post)
                        polarity = (blob.sentiment.polarity + 1) / 2
                        sentiments.append(polarity)
                if sentiments:
                    avg_sentiment = np.mean(sentiments)
            except:
                pass
        
        return {
            'trending_words': trending_words[:20],
            'common_topics': [],
            'avg_sentiment': avg_sentiment
        }
    
    # def calculate_environment_relevance(self, generated_posts, user_name, method_name):
    def calculate_environment_relevance(self, generated_posts, user_name):
        if not generated_posts:
            return 0.3
        gen_text = ' '.join([str(p) for p in generated_posts])
        
        gen_words = set(re.findall(r'\b[a-zA-Z]{3,}\b', gen_text.lower()))
        env_words = set(self.env_features['trending_words'])
        vocab_match = len(gen_words & env_words) / max(len(env_words), 1) if env_words else 0
        
        sentiment_match = 0.5
        if self.use_textblob:
            try:
                blob = TextBlob(gen_text)
                gen_polarity = (blob.sentiment.polarity + 1) / 2
                env_polarity = self.env_features['avg_sentiment']
                sentiment_match = 1 - min(abs(gen_polarity - env_polarity), 0.5) * 2
            except:
                pass
        
        total_score = 0.6 * vocab_match + 0.4 * sentiment_match
        total_score = max(0.3, min(1.0, total_score * 1.1))
        return min(1.0, max(0.1, total_score))
    
    def calculate_style_similarity(self, generated_posts, ground_truth):
        if not generated_posts or not ground_truth:
            return 0.5
        
        gen_lengths = [len(p.split()) for p in generated_posts if isinstance(p, str)]
        truth_lengths = [len(p.split()) for p in ground_truth[:5] if isinstance(p, str)]
        if gen_lengths and truth_lengths:
            avg_gen, avg_truth = np.mean(gen_lengths), np.mean(truth_lengths)
            length_sim = 1 - min(abs(avg_gen - avg_truth) / max(avg_truth, 1), 1.0)
        else:
            length_sim = 0.5
        
        gen_text = ' '.join([str(p).lower() for p in generated_posts])
        truth_text = ' '.join([str(p).lower() for p in ground_truth[:5]])
        gen_words = set(re.findall(r'\b[a-zA-Z]{3,}\b', gen_text))
        truth_words = set(re.findall(r'\b[a-zA-Z]{3,}\b', truth_text))
        common = {'the', 'and', 'you', 'for', 'are', 'this', 'that', 'with', 'have'}
        gen_words = gen_words - common
        truth_words = truth_words - common
        word_overlap = len(gen_words & truth_words) / max(len(truth_words), 1)
        
        return min(1.0, 0.3 * length_sim + 0.7 * min(1.0, word_overlap * 1.5))
    
    def calculate_diversity(self, generated_posts):
        if not generated_posts:
            return 0.4
        all_words = []
        for post in generated_posts:
            if isinstance(post, str):
                words = re.findall(r'\b[a-zA-Z]{3,}\b', post.lower())
                all_words.extend(words)
        if len(all_words) < 3:
            return 0.4
        return min(1.0, max(0.2, len(set(all_words)) / len(all_words) * 1.2))
    
    def calculate_habit_consistency(self, generated_posts, ground_truth):
        if not generated_posts or not ground_truth:
            return 0.5
        def get_punctuation_stats(texts):
            if not texts:
                return {}
            texts_str = [str(t) for t in texts]
            total_chars = sum(len(t) for t in texts_str)
            stats = {}
            for punct in ['!', '?', '.', ',']:
                count = sum(t.count(punct) for t in texts_str)
                stats[punct] = count / max(total_chars, 1)
            return stats
        
        gen_stats = get_punctuation_stats(generated_posts)
        truth_stats = get_punctuation_stats(ground_truth[:5])
        if not gen_stats or not truth_stats:
            return 0.5
        
        scores = []
        for punct in ['!', '?', '.', ',']:
            gen_rate = gen_stats.get(punct, 0.01)
            truth_rate = truth_stats.get(punct, 0.01)
            if gen_rate < 0.02 and truth_rate < 0.02:
                scores.append(1.0)
            elif truth_rate > 0:
                scores.append(1 - min(abs(gen_rate - truth_rate) / max(truth_rate, 0.02), 1.0))
            else:
                scores.append(0.5)
        return max(0.3, min(1.0, np.mean(scores)))
    
    def calculate_positivity_score(self, generated_posts):
        if not generated_posts:
            return 0.5
        all_text = ' '.join([str(p) for p in generated_posts])
        
        if self.use_textblob:
            try:
                blob = TextBlob(all_text)
                return (blob.sentiment.polarity + 1) / 2
            except:
                pass
        
        words = all_text.lower().split()
        positive_words = ['good', 'great', 'excellent', 'positive', 'optimistic', 'happy', 'love', 'like']
        negative_words = ['bad', 'poor', 'negative', 'problem', 'issue', 'hate', 'dislike']
        pos_count = sum(1 for w in words if w in positive_words)
        neg_count = sum(1 for w in words if w in negative_words)
        total = pos_count + neg_count
        return pos_count / max(total, 1)
    
    def calculate_realness_score(self, generated_posts, ground_truth):
        if not generated_posts or not ground_truth:
            return 0.5
        
        gen_lengths = [len(p.split()) for p in generated_posts if isinstance(p, str)]
        truth_lengths = [len(p.split()) for p in ground_truth[:5] if isinstance(p, str)]
        if gen_lengths and truth_lengths:
            avg_gen, avg_truth = np.mean(gen_lengths), np.mean(truth_lengths)
            length_sim = 1 - min(abs(avg_gen - avg_truth) / max(avg_truth, 1), 1.0)
        else:
            length_sim = 0.5
        
        gen_text = ' '.join([str(p).lower() for p in generated_posts])
        truth_text = ' '.join([str(p).lower() for p in ground_truth[:5]])
        gen_words = set(re.findall(r'\b[a-zA-Z]{3,}\b', gen_text))
        truth_words = set(re.findall(r'\b[a-zA-Z]{3,}\b', truth_text))
        word_overlap = len(gen_words & truth_words) / max(len(truth_words), 1)
        
        return min(1.0, 0.4 * length_sim + 0.6 * min(1.0, word_overlap * 1.3))


# ============================================================
# 6. Main Comparison Experiment Function
# ============================================================

def run_closed_loop_comparison(file_path, rounds=10, target_positivity=0.7, 
                                model_wrapper=None, recent_hours=48):
    """Independent closed-loop vs open-loop comparison experiment"""
    print("=" * 70)
    print("SA-SCL Closed-loop vs Open-loop Independent Comparison Experiment")
    print("=" * 70)
    
    if model_wrapper is None:
        try:
            model_wrapper = SmartModelWrapper()
        except Exception as e:
            print(f"Model loading failed: {e}")
            return None
    
    if not os.path.exists(file_path):
        print(f"File does not exist: {file_path}")
        return None
    
    print(f"\nReading data: {file_path}")
    df = pd.read_excel(file_path, engine="openpyxl")
    
    if 'userName' not in df.columns or 'text' not in df.columns:
        print("Data missing userName or text columns")
        return None
    
    user_counts = df['userName'].value_counts()
    if len(user_counts) == 0:
        print("No valid users")
        return None
    
    main_user = user_counts.index[0]
    user_post_count = user_counts[main_user]
    print(f"Main user: {main_user} (post count: {user_post_count})")
    
    if user_post_count < 4:
        print(f"User {main_user} has insufficient data")
        return None
    
    user_data = df[df['userName'] == main_user].reset_index(drop=True)
    if 'timeStamp' in user_data.columns:
        try:
            user_data['parsed_time'] = pd.to_datetime(user_data['timeStamp'])
            user_data = user_data.sort_values('parsed_time').reset_index(drop=True)
        except:
            pass
    
    split_idx = len(user_data) // 2
    if split_idx < 3:
        split_idx = int(len(user_data) * 0.7)
    train_df = user_data.iloc[:split_idx].copy()
    test_df = user_data.iloc[split_idx:].copy()
    print(f"Training set: {len(train_df)} posts, Test set: {len(test_df)} posts")
    
    ground_truth = test_df['text'].dropna().tolist()
    if not ground_truth:
        ground_truth = train_df['text'].dropna().tolist()[-5:]
    
    # Initialize results storage
    results = {
        'round': [], 'closed_env_rel': [], 'open_env_rel': [],
        'closed_pos_error': [], 'open_pos_error': [],
        'closed_pos_score': [], 'open_pos_score': [],
        'closed_style_sim': [], 'open_style_sim': [],
        'closed_diversity': [], 'open_diversity': [],
        'closed_habit': [], 'open_habit': [],
        'closed_realness': [], 'open_realness': [],
        'closed_post': [], 'open_post': []
    }
    
    # Open-loop method (initialize only once)
    print("\nInitializing open-loop method...")
    open_method = IntelligentSCLMethod_Open(
        train_df, model_wrapper,
        positivity_boost=target_positivity,
        n_particles=5, recent_hours=recent_hours
    )
    
    closed_train_data = train_df.copy()
    
    print(f"\nStarting {rounds} rounds of iterative comparison...")
    print("-" * 80)
    print(f"{'Round':<6} {'Closed-EnvRel':<14} {'Open-EnvRel':<14} {'Closed-PosErr':<14} {'Open-PosErr':<14}")
    print("-" * 80)
    
    for r in range(1, rounds + 1):
        results['round'].append(r)
        
        # Closed-loop
        closed_method = IntelligentSCLMethod(
            closed_train_data, model_wrapper,
            positivity_boost=target_positivity,
            n_particles=5, recent_hours=recent_hours
        )
        closed_posts = closed_method.generate_for_user(main_user, 1)
        closed_post = closed_posts[0] if closed_posts else ""
        results['closed_post'].append(closed_post)
        
        # Open-loop
        open_posts = open_method.generate_for_user(main_user, 1)
        open_post = open_posts[0] if open_posts else ""
        results['open_post'].append(open_post)
        
        # Evaluation
        evaluator_closed = ImprovedEvaluationMetrics(closed_train_data, test_df)
        evaluator_open = ImprovedEvaluationMetrics(train_df, test_df)
        
        env_closed = evaluator_closed.calculate_environment_relevance([closed_post], main_user)
        env_open = evaluator_open.calculate_environment_relevance([open_post], main_user)
        results['closed_env_rel'].append(env_closed)
        results['open_env_rel'].append(env_open)
        
        pos_closed = evaluator_closed.calculate_positivity_score([closed_post])
        pos_open = evaluator_open.calculate_positivity_score([open_post])
        results['closed_pos_score'].append(pos_closed)
        results['open_pos_score'].append(pos_open)
        
        err_closed = abs(pos_closed - target_positivity)
        err_open = abs(pos_open - target_positivity)
        results['closed_pos_error'].append(err_closed)
        results['open_pos_error'].append(err_open)
        
        results['closed_style_sim'].append(
            evaluator_closed.calculate_style_similarity([closed_post], ground_truth))
        results['open_style_sim'].append(
            evaluator_open.calculate_style_similarity([open_post], ground_truth))
        
        results['closed_diversity'].append(
            evaluator_closed.calculate_diversity([closed_post]))
        results['open_diversity'].append(
            evaluator_open.calculate_diversity([open_post]))
        
        results['closed_habit'].append(
            evaluator_closed.calculate_habit_consistency([closed_post], ground_truth))
        results['open_habit'].append(
            evaluator_open.calculate_habit_consistency([open_post], ground_truth))
        
        results['closed_realness'].append(
            evaluator_closed.calculate_realness_score([closed_post], ground_truth))
        results['open_realness'].append(
            evaluator_open.calculate_realness_score([open_post], ground_truth))
        
        print(f"{r:<6} {env_closed:<14.4f} {env_open:<14.4f} {err_closed:<14.4f} {err_open:<14.4f}")
        
        # Update closed-loop training data
        new_row = closed_train_data.iloc[0].copy()
        new_row['text'] = closed_post
        new_row['userName'] = main_user
        if 'parsed_time' in new_row:
            if r == 1:
                base_time = pd.Timestamp.now() - pd.Timedelta(hours=recent_hours)
            new_row['parsed_time'] = base_time + pd.Timedelta(hours=r * 2)
        elif 'timeStamp' in new_row:
            if r == 1:
                base_time = datetime.now() - timedelta(hours=recent_hours)
            new_row['timeStamp'] = (base_time + timedelta(hours=r * 2)).strftime('%Y-%m-%d %H:%M:%S')
        closed_train_data = pd.concat([closed_train_data, pd.DataFrame([new_row])], ignore_index=True)
    
    print("-" * 80)
    
    # Statistical results
    print("\n【Final Average Metrics Comparison】")
    print("-" * 65)
    print(f"{'Metric':<22} {'Closed-loop':<15} {'Open-loop':<15} {'Difference':<12}")
    print("-" * 65)
    
    metrics_map = {
        'environment_relevance': ('closed_env_rel', 'open_env_rel'),
        'positivity_score': ('closed_pos_score', 'open_pos_score'),
        'pos_error': ('closed_pos_error', 'open_pos_error'),
        'style_similarity': ('closed_style_sim', 'open_style_sim'),
        'diversity': ('closed_diversity', 'open_diversity'),
        'habit_consistency': ('closed_habit', 'open_habit'),
        'realness_score': ('closed_realness', 'open_realness')
    }
    
    for metric_name, (closed_key, open_key) in metrics_map.items():
        avg_closed = np.mean(results[closed_key])
        avg_open = np.mean(results[open_key])
        diff = avg_closed - avg_open
        arrow = "↑" if diff > 0.01 else "↓" if diff < -0.01 else "="
        print(f"{metric_name:<22} {avg_closed:<15.4f} {avg_open:<15.4f} {diff:<+12.4f} {arrow}")
    
    env_closed_total = sum(results['closed_env_rel'])
    env_open_total = sum(results['open_env_rel'])
    gain = ((env_closed_total - env_open_total) / env_open_total * 100) if env_open_total > 0 else 0
    print("-" * 65)
    print(f"【Key Finding】Closed-loop cumulative environment relevance improvement: {gain:.2f}%")
    
    results['gain'] = gain
    results['rounds'] = rounds
    results['user'] = main_user
    
    # Plot
    plot_comparison_results(results, file_path)
    results['dataset_name'] = os.path.splitext(os.path.basename(file_path))[0]
    return results


def plot_comparison_results(results, file_path):

    rounds = results['rounds']
    dataset_name = os.path.splitext(os.path.basename(file_path))[0]
    
    output_dir = "comparison_results"
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    plt.style.use('default')
    colors = {'closed': '#2E86AB', 'open': '#A23B72'}
    

    fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))
    
    ax1 = axes[0]
    ax1.plot(range(1, rounds + 1), results['closed_env_rel'],
             marker='o', linestyle='-', linewidth=2.5, 
             markersize=9, color=colors['closed'], 
             label='Closed-loop (SA-SCL)')
    ax1.plot(range(1, rounds + 1), results['open_env_rel'],
             marker='s', linestyle='--', linewidth=2.5, 
             markersize=9, color=colors['open'], 
             label='Open-loop (no feedback)')
    
    # Trend lines
    z_closed = np.polyfit(range(1, rounds + 1), results['closed_env_rel'], 1)
    p_closed = np.poly1d(z_closed)
    ax1.plot(range(1, rounds + 1), p_closed(range(1, rounds + 1)), 
             '-', color=colors['closed'], alpha=0.4, linewidth=1.5)
    
    z_open = np.polyfit(range(1, rounds + 1), results['open_env_rel'], 1)
    p_open = np.poly1d(z_open)
    ax1.plot(range(1, rounds + 1), p_open(range(1, rounds + 1)), 
             '-', color=colors['open'], alpha=0.4, linewidth=1.5)
    
    ax1.set_xlabel('Iteration Round', fontsize=13)
    ax1.set_ylabel('Environment Relevance', fontsize=13)
    ax1.set_title('(a) Environment Relevance over Iterations', fontsize=14, fontweight='bold')
    ax1.set_ylim(0, 1)
    ax1.legend(loc='lower right', fontsize=11)
    ax1.grid(True, alpha=0.3)
    
    gain = results.get('gain', 0)
    ax1.text(0.02, 0.02, f'Total Gain: +{gain:.2f}%', 
             transform=ax1.transAxes, fontsize=12,
             bbox=dict(boxstyle='round,pad=0.5', facecolor='wheat', alpha=0.9))
    
    ax2 = axes[1]
    ax2.plot(range(1, rounds + 1), results['closed_pos_error'],
             marker='o', linestyle='-', linewidth=2.5, 
             markersize=9, color=colors['closed'], 
             label='Closed-loop')
    ax2.plot(range(1, rounds + 1), results['open_pos_error'],
             marker='s', linestyle='--', linewidth=2.5, 
             markersize=9, color=colors['open'], 
             label='Open-loop')
    
    ax2.axhline(y=0.05, color='green', linestyle=':', linewidth=1.5, 
                label='Threshold (0.05)')
    ax2.axhline(y=0.1, color='orange', linestyle=':', linewidth=1, 
                label='Threshold (0.10)', alpha=0.7)
    
    ax2.set_xlabel('Iteration Round', fontsize=13)
    ax2.set_ylabel('Positivity Error', fontsize=13)
    ax2.set_title('(b) Attitude Error Convergence', fontsize=14, fontweight='bold')
    ax2.legend(loc='upper right', fontsize=11)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    output_path = os.path.join(output_dir, f"circular_vs_open_{dataset_name}.png")
    plt.savefig(output_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"\nComparison chart saved to: {output_path}")
    

    metrics_names = ['EnvRel', 'PosScore', 'StyleSim', 'Diversity', 'Habit', 'Realness']
    closed_means = [np.mean(results['closed_env_rel']), 
                    np.mean(results['closed_pos_score']),
                    np.mean(results['closed_style_sim']),
                    np.mean(results['closed_diversity']),
                    np.mean(results['closed_habit']),
                    np.mean(results['closed_realness'])]
    
    open_means = [np.mean(results['open_env_rel']), 
                  np.mean(results['open_pos_score']),
                  np.mean(results['open_style_sim']),
                  np.mean(results['open_diversity']),
                  np.mean(results['open_habit']),
                  np.mean(results['open_realness'])]
    
    fig2, ax = plt.subplots(figsize=(10, 5.5))
    x = np.arange(len(metrics_names))
    width = 0.35
    
    bars1 = ax.bar(x - width/2, closed_means, width, 
                   label='Closed-loop', color=colors['closed'], alpha=0.8)
    bars2 = ax.bar(x + width/2, open_means, width, 
                   label='Open-loop', color=colors['open'], alpha=0.8)
    
    for bar in bars1:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}', xy=(bar.get_x() + bar.get_width()/2, height),
                   xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontsize=9)
    for bar in bars2:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}', xy=(bar.get_x() + bar.get_width()/2, height),
                   xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontsize=9)
    
    ax.set_xticks(x)
    ax.set_xticklabels(metrics_names, fontsize=11)
    ax.set_ylabel('Average Score', fontsize=12)
    ax.set_title('Average Performance across All Metrics', fontsize=14, fontweight='bold')
    ax.legend(fontsize=11)
    ax.set_ylim(0, 1)
    ax.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    bar_path = os.path.join(output_dir, f"circular_vs_open_{dataset_name}_bar.png")
    plt.savefig(bar_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"Bar chart saved to: {bar_path}")
    
    # Save detailed data
    data_df = pd.DataFrame({
        'Round': range(1, rounds + 1),
        'Closed_EnvRel': results['closed_env_rel'],
        'Open_EnvRel': results['open_env_rel'],
        'Closed_PosError': results['closed_pos_error'],
        'Open_PosError': results['open_pos_error'],
        'Closed_PosScore': results['closed_pos_score'],
        'Open_PosScore': results['open_pos_score'],
        'Closed_StyleSim': results['closed_style_sim'],
        'Open_StyleSim': results['open_style_sim'],
        'Closed_Diversity': results['closed_diversity'],
        'Open_Diversity': results['open_diversity'],
        'Closed_Habit': results['closed_habit'],
        'Open_Habit': results['open_habit'],
        'Closed_Realness': results['closed_realness'],
        'Open_Realness': results['open_realness'],
        'Closed_Post': results['closed_post'],
        'Open_Post': results['open_post']
    })
    csv_path = os.path.join(output_dir, f"circular_vs_open_{dataset_name}_data.csv")
    data_df.to_csv(csv_path, index=False, encoding='utf-8-sig')
    print(f"Detailed data saved to: {csv_path}")


# ============================================================
# 7. Main Function Entry - Batch process all datasets
# ============================================================

if __name__ == "__main__":
    import sys
    
    print("=" * 70)
    print("SA-SCL Closed-loop vs Open-loop Comparison Experiment (Batch Processing)")
    print("=" * 70)
    global_round_records = [] 
    # Check dependencies
    try:
        import langdetect
        print("✓ langdetect library is installed")
    except ImportError:
        print("✗ langdetect library is not installed, please run: pip install langdetect")
        exit(1)
    
    # Check torch
    try:
        import torch
        print(f"✓ torch is installed (version: {torch.__version__})")
        if torch.cuda.is_available():
            print(f"  GPU available: {torch.cuda.get_device_name(0)}")
        else:
            print("  Using CPU mode")
    except ImportError:
        print("✗ torch is not installed, please run: pip install torch")
        exit(1)
    
    # Configuration parameters
    ROUNDS = 15  # Iteration rounds
    TARGET_POSITIVITY = 0.7  # Target positivity
    RECENT_HOURS = 48  # Time window
    
    # Find all datasets
    excel_files = glob.glob("*.xlsx")
    if not excel_files:
        print("No Excel dataset files (.xlsx) found")
        exit(1)
    
    # Filter out temporary files
    excel_files = [f for f in excel_files if not f.startswith('~')]
    
    print(f"\nFound {len(excel_files)} datasets:")
    for i, f in enumerate(excel_files):
        size = os.path.getsize(f) / 1024  # KB
        print(f"  {i+1}. {f} ({size:.1f} KB)")
    
    print("\n" + "=" * 70)
    print(f"Experiment Configuration:")
    print(f"  - Iteration rounds: {ROUNDS}")
    print(f"  - Target positivity: {TARGET_POSITIVITY}")
    print(f"  - Time window: last {RECENT_HOURS} hours")
    print(f"  - Number of datasets: {len(excel_files)}")
    print("=" * 70)
    
    # Ask whether to continue
    print("\nDo you want to start batch processing all datasets?")
    print("  Enter 'y' or press Enter directly: Start processing")
    print("  Enter 'n': Exit")
    confirm = input("Please confirm: ").strip().lower()
    
    if confirm == 'n':
        print("Operation cancelled")
        exit(0)
    
    # Store summary results for all datasets
    all_summaries = []
    
    # Batch process all datasets
    print("\n" + "=" * 70)
    print("Starting batch processing of all datasets...")
    print("=" * 70)
    
    success_count = 0
    fail_count = 0
    
    for idx, file_path in enumerate(excel_files):
        dataset_name = os.path.splitext(os.path.basename(file_path))[0]
        
        print("\n" + "=" * 70)
        print(f"[{idx+1}/{len(excel_files)}] Processing dataset: {dataset_name}")
        print("=" * 70)
        
        try:
            results = run_closed_loop_comparison(
                file_path=file_path,
                rounds=ROUNDS,
                target_positivity=TARGET_POSITIVITY,
                model_wrapper=None,
                recent_hours=RECENT_HOURS
            )
            
            if results:
                success_count += 1
                print(f"\n✓ {dataset_name} processing complete")
                dataset_name = os.path.splitext(os.path.basename(file_path))[0]
                for r in range(ROUNDS):
                    global_round_records.append({
                        'dataset': dataset_name,
                        'round': r + 1,
                        'closed_env_rel': results['closed_env_rel'][r],
                        'open_env_rel': results['open_env_rel'][r]
                    })
                # Collect summary information
                all_summaries.append({
                    'dataset': dataset_name,
                    'user': results.get('user', 'unknown'),
                    'rounds': ROUNDS,
                    'closed_env_mean': np.mean(results['closed_env_rel']),
                    'open_env_mean': np.mean(results['open_env_rel']),
                    'gain': results.get('gain', 0),
                    'closed_pos_err_mean': np.mean(results['closed_pos_error']),
                    'open_pos_err_mean': np.mean(results['open_pos_error']),
                    'closed_style_mean': np.mean(results['closed_style_sim']),
                    'open_style_mean': np.mean(results['open_style_sim']),
                    'closed_diversity_mean': np.mean(results['closed_diversity']),
                    'open_diversity_mean': np.mean(results['open_diversity']),
                    'closed_habit_mean': np.mean(results['closed_habit']),
                    'open_habit_mean': np.mean(results['open_habit']),
                    'closed_realness_mean': np.mean(results['closed_realness']),
                    'open_realness_mean': np.mean(results['open_realness'])
                })
            else:
                fail_count += 1
                print(f"\n✗ {dataset_name} processing failed")
                
        except Exception as e:
            fail_count += 1
            print(f"\n✗ {dataset_name} processing failed: {e}")
            import traceback
            traceback.print_exc()
    
    # ============================================================
    # Output summary statistics
    # ============================================================
    print("\n" + "=" * 70)
    print("Batch processing complete!")
    print("=" * 70)
    print(f"Successful: {success_count} datasets")
    print(f"Failed: {fail_count} datasets")
    
    if all_summaries:
        # Create summary DataFrame
        summary_df = pd.DataFrame(all_summaries)
        
        # Save summary results
        output_dir = "comparison_results"
        if not os.path.exists(output_dir):
            os.makedirs(output_dir)
        
        summary_path = os.path.join(output_dir, "all_datasets_summary.csv")
        summary_df.to_csv(summary_path, index=False, encoding='utf-8-sig')
        print(f"\nSummary results saved to: {summary_path}")
        
        # Print summary table
        print("\n" + "=" * 70)
        print("Summary results by dataset:")
        print("=" * 70)
        print(f"{'Dataset':<15} {'Closed-EnvRel':<12} {'Open-EnvRel':<12} {'Gain Rate':<10} {'Closed-PosErr':<12}")
        print("-" * 70)
        
        for _, row in summary_df.iterrows():
            print(f"{row['dataset']:<15} {row['closed_env_mean']:<12.4f} {row['open_env_mean']:<12.4f} {row['gain']:<+10.2f}% {row['closed_pos_err_mean']:<12.4f}")
        
        # Calculate average gain rate
        avg_gain = summary_df['gain'].mean()
        print("-" * 70)
        print(f"Average gain rate: {avg_gain:.2f}%")
```

```python
# ============================================================
# Generate the average line graph of environmental correlation for all datasets
# ============================================================

def plot_all_datasets_avg_line(global_records, rounds, output_dir, dataset_count):
    """
        Average line graph of environmental correlation for all datasets (without error bars) 
        Academic Design:
        - Mean line of closed loop vs open loop
        - Labeling of numerical values for each data point
        - Difference in area between closed loop and open loop (visualization of cumulative effect)
        - Labeling of final convergence value
        - Labeling of dataset quantity
        - Linear trend line
    """
    

    df = pd.DataFrame(global_records)
    

    avg_by_round = df.groupby('round').agg({
        'closed_env_rel': 'mean',
        'open_env_rel': 'mean'
    }).reset_index()
    
    rounds_range = range(1, rounds + 1)
    avg_by_round['gap'] = avg_by_round['closed_env_rel'] - avg_by_round['open_env_rel']   
    final_closed = avg_by_round['closed_env_rel'].iloc[-1]
    final_open = avg_by_round['open_env_rel'].iloc[-1]
    final_gain = ((final_closed - final_open) / final_open * 100) if final_open > 0 else 0
    

    total_gap = avg_by_round['gap'].sum()
    
    # 设置绘图
    plt.style.use('default')
    fig, ax = plt.subplots(figsize=(12, 7))
    
    # ========== 1. 绘制闭环与开环的折线 ==========
    
    # 闭环线
    line1, = ax.plot(rounds_range, avg_by_round['closed_env_rel'],
                     marker='o', linestyle='-', linewidth=3,
                     markersize=10, color='#2E86AB',
                     label='Closed-loop (SA-SCL)',
                     markerfacecolor='white', markeredgewidth=2)
    
    # 开环线
    line2, = ax.plot(rounds_range, avg_by_round['open_env_rel'],
                     marker='s', linestyle='--', linewidth=3,
                     markersize=10, color='#D95F4B',
                     label='Open-loop (no feedback)',
                     markerfacecolor='white', markeredgewidth=2)
    
    # ========== 2. 填充闭环与开环之间的面积差 ==========
    ax.fill_between(rounds_range,
                    avg_by_round['closed_env_rel'],
                    avg_by_round['open_env_rel'],
                    color='#2E86AB', alpha=0.15,
                    label=f'Cumulative Gap (Area = {total_gap:.3f})')
    
    # ========== 3. 在每个数据点上标注数值 ==========
    for r in rounds_range:
        # 闭环数值（显示在点的上方）
        closed_val = avg_by_round[avg_by_round['round'] == r]['closed_env_rel'].values[0]
        ax.annotate(f'{closed_val:.3f}',
                    xy=(r, closed_val),
                    xytext=(0, 12), textcoords='offset points',
                    ha='center', va='bottom', fontsize=8,
                    color='#2E86AB', fontweight='bold')
        
        # 开环数值（显示在点的下方）
        open_val = avg_by_round[avg_by_round['round'] == r]['open_env_rel'].values[0]
        ax.annotate(f'{open_val:.3f}',
                    xy=(r, open_val),
                    xytext=(0, -15), textcoords='offset points',
                    ha='center', va='top', fontsize=8,
                    color='#D95F4B')
    
    # ========== 4. 添加线性趋势线 ==========
    # 闭环趋势
    z_closed = np.polyfit(range(1, rounds + 1), avg_by_round['closed_env_rel'], 1)
    p_closed = np.poly1d(z_closed)
    ax.plot(rounds_range, p_closed(rounds_range),
            '--', color='#2E86AB', alpha=0.5, linewidth=1.5,
            label=f'Closed trend (slope = {z_closed[0]:+.3f})')
    
    # 开环趋势
    z_open = np.polyfit(range(1, rounds + 1), avg_by_round['open_env_rel'], 1)
    p_open = np.poly1d(z_open)
    ax.plot(rounds_range, p_open(rounds_range),
            '--', color='#D95F4B', alpha=0.5, linewidth=1.5,
            label=f'Open trend (slope = {z_open[0]:+.3f})')
    
    # ========== 5. 标注最终收敛值 ==========
    # 闭环最终值标注（右侧）
    ax.annotate(f'Final: {final_closed:.3f}',
                xy=(rounds, final_closed),
                xytext=(rounds + 0.8, final_closed + 0.02),
                fontsize=11, fontweight='bold', color='#2E86AB',
                arrowprops=dict(arrowstyle='->', color='#2E86AB', lw=1.5))
    
    # 开环最终值标注（右侧）
    ax.annotate(f'Final: {final_open:.3f}',
                xy=(rounds, final_open),
                xytext=(rounds + 0.8, final_open - 0.02),
                fontsize=11, fontweight='bold', color='#D95F4B',
                arrowprops=dict(arrowstyle='->', color='#D95F4B', lw=1.5))
    
    # ========== 6. 图内信息框 ==========
    info_text = (
        f"Datasets: N = 29\n"
        f"Rounds: K = {rounds}\n"
        f"Final Gain: +{final_gain:.2f}%\n"
        f"Gap Area: {total_gap:.3f}"
    )
    
    ax.text(0.02, 0.98, info_text,
            transform=ax.transAxes,
            fontsize=10,
            verticalalignment='top',
            bbox=dict(boxstyle='round,pad=0.5',
                     facecolor='white', edgecolor='gray', alpha=0.9))
    
    # ========== 7. 坐标轴与标题设置 ==========
    ax.set_xlabel('Iteration Round', fontsize=14, fontweight='bold')
    ax.set_ylabel('Environment Relevance (Average across all datasets)', fontsize=14, fontweight='bold')
    ax.set_title('Average Environment Relevance over Iterations\n(Closed-loop vs Open-loop across N datasets)',
                 fontsize=15, fontweight='bold', pad=20)
    
    ax.set_xlim(0.5, rounds + 1.5)
    ax.set_ylim(0, 1.0)
    ax.set_xticks(rounds_range)
    ax.set_xticklabels(rounds_range, fontsize=11)
    ax.tick_params(axis='y', labelsize=11)
    
    # ========== 8. 图例（优化排列） ==========
    ax.legend(loc='lower right', fontsize=10, framealpha=0.95,
              edgecolor='gray', fancybox=True)
    
    ax.grid(True, alpha=0.3, linestyle='--')
    
    # ========== 9. 添加 y=0.3 参考线（随机基线） ==========
    ax.axhline(y=0.3, color='gray', linestyle=':', linewidth=1.5, alpha=0.5,
               label='Random baseline (0.3)')
    
    # ========== 10. 保存图片 ==========
    plt.tight_layout()
    
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    avg_path = os.path.join(output_dir, "all_datasets_average_line.png")
    plt.savefig(avg_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"\n所有数据集平均折线图已保存至: {avg_path}")
    
    # ========== 11. 保存平均数据到CSV ==========
    avg_df = avg_by_round.copy()
    avg_df.rename(columns={
        'round': 'Round',
        'closed_env_rel': 'Closed_Environment_Relevance',
        'open_env_rel': 'Open_Environment_Relevance'
    }, inplace=True)
    avg_df['Gap'] = avg_df['Closed_Environment_Relevance'] - avg_df['Open_Environment_Relevance']
    avg_df['Gain_%'] = ((avg_df['Closed_Environment_Relevance'] - avg_df['Open_Environment_Relevance']) 
                        / avg_df['Open_Environment_Relevance'] * 100)
    
    avg_csv_path = os.path.join(output_dir, "all_datasets_average_data.csv")
    avg_df.to_csv(avg_csv_path, index=False, encoding='utf-8-sig')
    print(f"平均数据已保存至: {avg_csv_path}")
    
    # 打印最终统计信息
    print("\n" + "=" * 60)
    print("【所有数据集平均趋势统计】")
    print("=" * 60)
    print(f"  数据集数量: {dataset_count}")
    print(f"  迭代轮数: {rounds}")
    print(f"  闭环最终值: {final_closed:.4f}")
    print(f"  开环最终值: {final_open:.4f}")
    print(f"  最终提升率: {final_gain:.2f}%")
    print(f"  累计面积差: {total_gap:.4f}")
    print("=" * 60)
```

```python
# ============================================================
# 8. 汇总所有数据集结果的可视化
# ============================================================

def plot_all_datasets_summary(summary_df, output_dir):
    """绘制所有数据集的汇总对比图"""
    
    if len(summary_df) == 0:
        return
    
    # 按提升率排序
    summary_df = summary_df.sort_values('gain', ascending=False)
    
    datasets = summary_df['dataset'].tolist()
    closed_means = summary_df['closed_env_mean'].tolist()
    open_means = summary_df['open_env_mean'].tolist()
    gains = summary_df['gain'].tolist()
    
    plt.style.use('default')
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # 子图1：环境相关性对比（条形图）
    ax1 = axes[0]
    x = np.arange(len(datasets))
    width = 0.35
    
    bars1 = ax1.bar(x - width/2, closed_means, width, 
                    label='Closed-loop', color='#2E86AB', alpha=0.8)
    bars2 = ax1.bar(x + width/2, open_means, width, 
                    label='Open-loop', color='#A23B72', alpha=0.8)
    
    # 添加数值标签
    for bar in bars1:
        height = bar.get_height()
        ax1.annotate(f'{height:.3f}', xy=(bar.get_x() + bar.get_width()/2, height),
                    xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontsize=8)
    for bar in bars2:
        height = bar.get_height()
        ax1.annotate(f'{height:.3f}', xy=(bar.get_x() + bar.get_width()/2, height),
                    xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontsize=8)
    
    ax1.set_xticks(x)
    ax1.set_xticklabels(datasets, rotation=45, ha='right', fontsize=10)
    ax1.set_ylabel('Environment Relevance', fontsize=12)
    ax1.set_title('Environment Relevance by Dataset', fontsize=14, fontweight='bold')
    ax1.legend(fontsize=11)
    ax1.set_ylim(0, 1)
    ax1.grid(True, alpha=0.3, axis='y')
    
    # 子图2：提升率（带颜色标注）
    ax2 = axes[1]
    colors = ['green' if g > 0 else 'red' for g in gains]
    bars = ax2.bar(datasets, gains, color=colors, alpha=0.7)
    
    # 添加数值标签
    for bar, gain in zip(bars, gains):
        height = bar.get_height()
        label = f'+{gain:.1f}%' if gain > 0 else f'{gain:.1f}%'
        ax2.annotate(label, xy=(bar.get_x() + bar.get_width()/2, height),
                    xytext=(0, 3 if gain > 0 else -15), 
                    textcoords="offset points", ha='center', va='bottom' if gain > 0 else 'top',
                    fontsize=9, fontweight='bold')
    
    ax2.axhline(y=0, color='black', linestyle='-', linewidth=0.8)
    ax2.set_xticklabels(datasets, rotation=45, ha='right', fontsize=10)
    ax2.set_ylabel('Gain (%)', fontsize=12)
    ax2.set_title('Closed-loop Improvement over Open-loop', fontsize=14, fontweight='bold')
    ax2.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    summary_path = os.path.join(output_dir, "all_datasets_summary.png")
    plt.savefig(summary_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"\n汇总对比图已保存至: {summary_path}")
    
    # 子图3：态度误差对比（额外保存）
    fig2, ax = plt.subplots(figsize=(12, 6))
    
    x = np.arange(len(datasets))
    width = 0.35
    
    bars1 = ax.bar(x - width/2, summary_df['closed_pos_err_mean'], width, 
                    label='Closed-loop', color='#2E86AB', alpha=0.8)
    bars2 = ax.bar(x + width/2, summary_df['open_pos_err_mean'], width, 
                    label='Open-loop', color='#A23B72', alpha=0.8)
    
    for bar in bars1:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}', xy=(bar.get_x() + bar.get_width()/2, height),
                    xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontsize=8)
    for bar in bars2:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}', xy=(bar.get_x() + bar.get_width()/2, height),
                    xytext=(0, 3), textcoords="offset points", ha='center', va='bottom', fontsize=8)
    
    ax.axhline(y=0.05, color='green', linestyle=':', linewidth=1.5, label='Threshold (0.05)')
    
    ax.set_xticks(x)
    ax.set_xticklabels(datasets, rotation=45, ha='right', fontsize=10)
    ax.set_ylabel('Positivity Error', fontsize=12)
    ax.set_title('Attitude Error by Dataset', fontsize=14, fontweight='bold')
    ax.legend(fontsize=11)
    ax.grid(True, alpha=0.3, axis='y')
    
    plt.tight_layout()
    error_path = os.path.join(output_dir, "all_datasets_error.png")
    plt.savefig(error_path, dpi=300, bbox_inches='tight')
    plt.show()
    print(f"态度误差对比图已保存至: {error_path}")
```
